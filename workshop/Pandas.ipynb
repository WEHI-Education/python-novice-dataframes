## Series

<div style="background:#e3f2fd;border-left:6px solid #0096ff;padding:6px 12px;border-radius:4px;color:#000;">
🔹 A <code>Series</code> is a one-dimensional labelled array: the building block of every DataFrame column.
</div>

A `Series` is a one-dimensional array-like object containing a sequence of values and an associated array of row labels, called its *index*.

You can recognise a `Series` because two attributes are printed at the bottom: the `Name` (if it has one) and the `dtype` (data type of the values).

### Creating a Series

<div style="background:#e3f2fd;border-left:6px solid #0096ff;padding:6px 12px;border-radius:4px;color:#000;">
🔹 Build a Series from a list or a dict, with or without labels.
</div>

The simplest `Series` is formed from a list of values. Let's record the weight (kg) of three people:

```python
weights = pd.Series([68, 83, 112])
weights
```

The string representation of a `Series` shows the index on the left and the values on the right. When we don't supply an index, pandas creates one for us: the integers `0, 1, 2, ...`.

You can get the array of values and the index object via the `values` and `index` attributes.

```python
weights.values
```

```python
weights.index  # like range(3)
```

```python
list(weights.index)  # if you want the index as a plain list
```

Usually we want an index that identifies each data point with a meaningful label. We can also give the `Series` a `name`.

```python
weights = pd.Series([68, 83, 112], index=["alice", "bob", "charles"], name="weight")
weights
```

```python
list(weights.index)
```

You can also create a `Series` from a `dict`. The keys become the index labels. Here are heights (cm) for four people:

```python
heights = pd.Series({"alice": 172, "bob": 181, "charles": 185, "darwin": 178}, name="height")
heights
```

### Selecting and filtering in Series

<div style="background:#e3f2fd;border-left:6px solid #0096ff;padding:6px 12px;border-radius:4px;color:#000;">
🔹 Pick values by label, by position, or with a boolean condition.
</div>

Compared with `NumPy` arrays, you can use the labels in the index to select single values or a set of values.

```python
weights
```

```python
weights["alice"]
```

```python
weights[["bob", "charles"]]
# Here ["bob", "charles"] is interpreted as a list of index labels
```

You can also access items by integer *position*, like in a regular array. Positions start at **0**.

To make it clear whether you are selecting by label or by position, always use the `loc` attribute for labels and the `iloc` attribute for positions.

```python
weights.loc["alice"]   # by label
```

```python
weights.iloc[0]        # by position
```

```python
weights.iloc[[0, 2]]   # first and third items
```

Passing a boolean condition returns only the items where the condition is `True`. This is called *boolean filtering* and you will use it constantly.

```python
weights > 70
```

```python
weights[weights > 70]
```

Slicing with labels behaves differently from normal Python slicing: the end point is **inclusive**. Slicing by position is exclusive, as usual.

```python
weights.loc["alice":"bob"]   # includes bob
```

```python
weights.iloc[0:2]            # positions 0 and 1
```

Slicing a `Series` also slices its index labels. This can lead to surprises when the labels are the default integers, so be careful:

```python
surprise = pd.Series([68, 83, 112, 95])
surprise_slice = surprise.iloc[2:]
surprise_slice
```

Oh look! The first element of the slice has index label **2**. The labels 0 and 1 are gone with the rows that were sliced away.

So `surprise_slice.loc[2]` and `surprise_slice.iloc[0]` return the *same* element, and `surprise_slice.loc[0]` would fail with a `KeyError`. This is another reason to always be explicit with `loc` and `iloc`.

```python
surprise_slice.loc[2]
```

```python
surprise_slice.iloc[0]
```

### Operations on Series

<div style="background:#e3f2fd;border-left:6px solid #0096ff;padding:6px 12px;border-radius:4px;color:#000;">
🔹 Maths on a Series is elementwise, broadcasts scalars, and aligns on the index.
</div>

`Series` objects behave much like a one-dimensional `ndarray`, and you can pass them to `NumPy` functions.

```python
np.log(weights)
```

Arithmetic operations apply *elementwise*. If you combine a `Series` with a single number, that number is applied to every item. This is called *broadcasting*.

```python
weights + 5
```

```python
weights * [1, 2, 3]   # elementwise with a list of the same length
```

The same is true for comparison operators, which is what made the boolean filtering above work.

```python
weights > 80
```

When you combine *two* `Series`, pandas first **aligns** them on their index labels and then applies the operation. Labels present in only one `Series` get a missing value, shown as `NaN` (not a number).

Let's compute body mass index (weight / height²) from our two Series. Note that `heights` has an extra person, darwin:

```python
bmi = weights / (heights / 100) ** 2
bmi
```

**Other ways to create a DataFrame**

You can also create a `DataFrame` by passing a `dict` of `Series` objects. This is the `people` DataFrame we will use for the rest of Section 1:

```python
people_dict = {
    "weight":    pd.Series([68, 83, 112], index=["alice", "bob", "charles"]),
    "birthyear": pd.Series([1984, 1985, 1992], index=["bob", "alice", "charles"], name="year"),
    "children":  pd.Series([0, 3], index=["charles", "bob"]),
    "hobby":     pd.Series(["Biking", "Dancing"], index=["alice", "bob"]),
}
people = pd.DataFrame(people_dict)
people
```

A few things to note:
* The `Series` were automatically **aligned** on their index (bob's birthyear ended up in bob's row even though it was listed first).
* Missing values are represented as `NaN`.
* `Series` names are ignored (the name `"year"` was dropped in favour of the dict key).

If you pass a list of columns and/or index row labels to the `DataFrame` constructor, it guarantees that exactly these columns and/or rows will exist, in that order, and no others.

```python
pd.DataFrame(
    people_dict,
    columns=["birthyear", "weight", "height"],
    index=["bob", "alice", "eugene"],
)
```

Another convenient way is to pass all the values as a list of lists (or an `ndarray`) and specify the column names and row labels separately:

```python
values = [
    [1985, np.nan, "Biking",   68],
    [1984, 3,      "Dancing",  83],
    [1992, 0,      np.nan,    112],
]
pd.DataFrame(
    values,
    columns=["birthyear", "children", "hobby", "weight"],
    index=["alice", "bob", "charles"],
)
```

**Don't forget to `.copy()`**

<div style="background:#fff8e1;border-left:6px solid #ffb300;padding:8px 12px;border-radius:4px;color:#000;">
<b>💡 Watch out:</b> in Python, <code>new_df = old_df</code> does <i>not</i> make a copy. It creates a second name (an "alias") for the same object, so changes through one name show up in the other.
</div>

Let's see this in action, and then fix it with the `copy()` method.

```python
people_alias = people                          # NOT a copy!
people_alias["nickname"] = ["Al", "Bobby", "Chuck"]
people                                         # people changed too
```

```python
people_copy = people.copy()                    # a real, independent copy
people_copy["nickname"] = "-"
people_copy
```

```python
people                                         # unchanged this time
```

```python
del people["nickname"]   # tidy up; more on removing columns shortly
people
```

### Methods

<div style="background:#e3f2fd;border-left:6px solid #0096ff;padding:6px 12px;border-radius:4px;color:#000;">

Extracting attributes is not the only use of the `object.xyz` dot syntax. The same syntax lets you call functions that belong to the object, using `object.method()`. Unlike R, there is no standalone `mean()` function that works on a DataFrame: you call the DataFrame's own `mean()` method.

Aggregation methods such as `mean()`, `sum()`, `min()` and `max()` work on each column and return a `Series`:

```python
people[["weight", "birthyear"]].mean()
```

What happens if we call `.mean()` on the whole DataFrame? The `hobby` column contains text, and pandas will not silently guess what the mean of some strings should be.

```python
try:
    people.mean()
except TypeError as e:
    print("TypeError:", e)
```

Tell pandas to only use the numeric columns with `numeric_only=True`.

```python
people.mean(numeric_only=True)
```

Text columns have their own useful methods. `value_counts()` tallies each distinct value:

```python
people["hobby"].value_counts()
```

### Adding and removing columns

<div style="background:#e3f2fd;border-left:6px solid #0096ff;padding:6px 12px;border-radius:4px;color:#000;">
🔹 Add columns like dict keys; remove them with <code>pop</code> or <code>del</code>.

You can generally treat a `DataFrame` like a dictionary of `Series`, so the following all work:

### Assigning new columns

<div style="background:#e3f2fd;border-left:6px solid #0096ff;padding:6px 12px;border-radius:4px;color:#000;">
🔹 <code>assign()</code> returns a new DataFrame with extra columns; chain it with <code>lambda</code>.
</div>

You can also create new columns by calling the `assign()` method. Note that this returns a *new* `DataFrame` and leaves the original unchanged.

```python
people.assign(
    body_mass_index = people["weight"] / (people["height"] / 100) ** 2,
    has_pets = people["pets"] > 0
)
```

Note that you cannot use a column created within the same `assign()` call:

```python
try:
    people.assign(
        body_mass_index = people["weight"] / (people["height"] / 100) ** 2,
        overweight = people["body_mass_index"] > 25
    )
except KeyError as e:
    print("KeyError:", e)
```

One fix is to split this into two consecutive assignments.

```python
df5 = people.assign(body_mass_index = people["weight"] / (people["height"] / 100) ** 2)
df5.assign(overweight = df5["body_mass_index"] > 25)
```

Having to create a temporary variable `df5` is not very convenient. You may want to just chain the `assign()` calls, but that does not work either, because `people` itself was not modified by the first call:

```python
try:
    (people
         .assign(body_mass_index = people["weight"] / (people["height"] / 100) ** 2)
         .assign(overweight = people["body_mass_index"] > 25)
    )
except KeyError as e:
    print("KeyError:", e)
```

But fear not, there is a simple solution. You can pass a function to `assign()` (typically a `lambda`), and it will be called with the *intermediate* `DataFrame` as its argument:

```python
(people
     .assign(body_mass_index = lambda df: df["weight"] / (df["height"] / 100) ** 2)
     .assign(overweight = lambda df: df["body_mass_index"] > 25)
)
```

Problem solved! This *method chaining* style is the pandas equivalent of the `|>` pipe in R.

## Loading data from CSV

<div style="background:#e3f2fd;border-left:6px solid #0096ff;padding:6px 12px;border-radius:4px;color:#000;">
🔹 <code>pd.read_csv()</code> turns a CSV file into a DataFrame.

Pandas can read tables from many file types, including CSV, Excel, JSON, HTML and parquet.

The data files live in the `data` folder next to this notebook. If you are running on Google Colab (or anywhere the folder is missing), the next cell downloads them from the GitHub repository first.

```python
import os
import urllib.request

# Check if the data directory exists; if not, create it and download the files
if not os.path.exists("data"):
    print("Data directory not found. Creating and downloading files...")
    os.makedirs("data")

    # Base URL for raw files from GitHub
    base_url = "https://raw.githubusercontent.com/WEHI-Education/python-novice-dataframes/main/workshop/data/"

    # List of files to download
    files = ["wdbc.names", "wdbc.data.csv", "wdbc.data"]

    for file in files:
        url = base_url + file
        local_path = f"data/{file}"
        try:
            print(f"Downloading {file}...")
            urllib.request.urlretrieve(url, local_path)
            print(f"✓ Successfully downloaded {file}")
        except Exception as e:
            print(f"✗ Error downloading {file}: {e}")

    print("Download complete!")
else:
    print("Data directory already exists.")
```

## Operations on DataFrame

<div style="background:#e3f2fd;border-left:6px solid #0096ff;padding:6px 12px;border-radius:4px;color:#000;">
🔹 NumPy functions, broadcasting and aggregation work across the whole table.
</div>

Although `DataFrame`s do not try to mimic `NumPy` arrays, there are a few similarities. Let's create a `DataFrame` of monthly scores for our people to demonstrate this:

```python
scores = pd.DataFrame(
    np.array([[8, 8, 9], [10, 9, 9], [4, 8, 2], [9, 10, 10]]),
    columns=["sep", "oct", "nov"],
    index=["alice", "bob", "charles", "darwin"],
)
scores
```

You can apply `NumPy` mathematical functions to a `DataFrame`. The function is applied to every value.

```python
np.sqrt(scores)
```

Adding a single value to a `DataFrame` adds it to every element. This is *broadcasting* again.

```python
scores + 1
```

The same is true for all other binary operations, including arithmetic (`*`, `/`, `**`...) and comparisons (`>`, `==`...).

```python
scores >= 5
```

Aggregation operations such as `max`, `sum` or `mean` apply to each column and return a `Series`.

```python
scores.mean()
```

The `all` method is also an aggregation: it checks whether all values are `True`. Let's see in which months everybody scored more than 5.

```python
(scores > 5).all()
```

Most of these methods take an `axis` argument. The default `axis=0` runs the operation down each column; `axis=1` runs it across each row. Let's find out who scored more than 5 every month:

```python
(scores > 5).all(axis=1)
```

The `any` method returns `True` if any value is `True`. Who scored a 10 at least once?

```python
(scores == 10).any(axis=1)
```

If you combine a `DataFrame` with a `Series`, pandas broadcasts the operation across all *rows*, matching the `Series` index to the column names. Let's subtract each month's mean from the scores:

```python
scores - scores.mean()   # equivalent to: scores - [7.75, 8.75, 7.50]
```

We subtracted `7.75` from all September scores, `8.75` from October and `7.50` from November. It is equivalent to subtracting this `DataFrame`:

```python
pd.DataFrame([[7.75, 8.75, 7.50]] * 4, index=scores.index, columns=scores.columns)
```

To subtract the *global* mean from every score, take the mean of the underlying array:

```python
scores - scores.values.mean()   # subtracts the global mean (8.00) from all scores
```

## Automatic alignment for DataFrames

<div style="background:#e3f2fd;border-left:6px solid #0096ff;padding:6px 12px;border-radius:4px;color:#000;">
🔹 Binary operations align on both row labels and column names; unmatched cells become <code>NaN</code>.
</div>

Just like `Series`, when you combine two `DataFrame`s pandas automatically aligns them by row index label *and* by column name. Let's create a `DataFrame` of bonus points, for slightly different people and months:

```python
bonus = pd.DataFrame(
    np.array([[0, np.nan, 2], [np.nan, 1, 0], [0, 1, 0], [3, 3, 0]]),
    columns=["oct", "nov", "dec"],
    index=["bob", "eugene", "darwin", "charles"],
)
bonus
```

```python
scores + bonus
```

The addition worked where both tables had a value, but far too many cells are now empty. When aligning, some rows and columns were only present on one side, so they were treated as missing (`NaN`) on the other, and adding `NaN` to a number gives `NaN`.

If a missing value should be treated as zero, use the `add()` method with `fill_value`. Only cells that are missing on *both* sides stay `NaN`:

```python
scores.add(bonus, fill_value=0)
```

Alternatively, use `dropna()` to discard rows or columns that are entirely `NaN` (`how="all"`):

```python
total = scores + bonus
total.dropna(how="all")               # drop rows that are all NaN
total.dropna(how="all", axis=1)       # drop columns that are all NaN
```

## Combining DataFrames

<div style="background:#e3f2fd;border-left:6px solid #0096ff;padding:6px 12px;border-radius:4px;color:#000;">
🔹 Two ways to bring tables together: joins on a key, or stacking with concat.

Real analyses usually involve more than one table. `pandas` gives you SQL-style joins with `merge()` and simple stacking with `concat()`.

## SQL like joins

<div style="background:#e3f2fd;border-left:6px solid #0096ff;padding:6px 12px;border-radius:4px;color:#000;">
🔹 <code>pd.merge()</code> joins two tables on a shared key: inner, outer, left or right.
</div>

One powerful feature of `pandas` is its ability to perform SQL-like joins. Inner joins, left/right outer joins and full outer joins are all supported. Let's create two small tables: where each person lives, and some facts about cities.

```python
residents = pd.DataFrame({
    "name":  ["alice", "bob", "charles", "darwin"],
    "city":  ["Melbourne", "Sydney", "Perth", "Melbourne"],
    "state": ["VIC", "NSW", "WA", "VIC"],
})
residents
```

```python
cities = pd.DataFrame({
    "city":  ["Melbourne", "Sydney", "Brisbane"],
    "state": ["Victoria", "New South Wales", "Queensland"],
    "lat":   [-37.81, -33.87, -27.47],
    "lng":   [144.96, 151.21, 153.03],
})
cities
```

Now let's join them on the `city` column using the `merge()` function:

```python
pd.merge(left=residents, right=cities, on="city")
```

Two things to note:
* Both tables have a column named `state`, so in the result they were renamed `state_x` and `state_y`. You can control the suffixes with the `suffixes` argument.
* charles (Perth) and Brisbane were dropped, because they do not appear in *both* tables. This is the equivalent of a SQL `INNER JOIN`, the default.

For a `FULL OUTER JOIN`, where nothing is dropped and the gaps are filled with `NaN`, specify `how="outer"`:

```python
pd.merge(left=residents, right=cities, on="city", how="outer", suffixes=("_abbrev", "_full"))
```

A `LEFT OUTER JOIN` (`how="left"`) keeps every row of the left table. `how="right"` keeps every row of the right table.

```python
pd.merge(left=residents, right=cities, on="city", how="left")
```

```python
pd.merge(left=residents, right=cities, on="city", how="right")
```

If the key columns have different names in the two tables, use `left_on` and `right_on`:

```python
cities_renamed = cities.rename(columns={"city": "city_name"})
pd.merge(left=residents, right=cities_renamed, left_on="city", right_on="city_name")
```

If the key is in the *index* of one or both tables, use `left_index=True` and/or `right_index=True`. Let's attach each person's city to the `people` DataFrame, whose index holds the names:

```python
pd.merge(left=people, right=residents.set_index("name"), left_index=True, right_index=True, how="left")
```

## Concatenation

<div style="background:#e3f2fd;border-left:6px solid #0096ff;padding:6px 12px;border-radius:4px;color:#000;">
🔹 <code>pd.concat()</code> stacks DataFrames vertically (new rows) or horizontally (new columns).
</div>

Rather than joining on a key, sometimes we just want to stack tables. Let's create a `DataFrame` with two more people and concatenate it with `people`:

```python
new_people = pd.DataFrame({
    "weight":  [95, 70],
    "height":  [178, 190],
    "hobby":   ["Running", np.nan],
    "age":     [48, 29],
    "over_30": [True, False],
    "pets":    [5, 0],
}, index=["darwin", "eugene"])
new_people
```

```python
everyone = pd.concat([people, new_people])
everyone
```

The tables were aligned by column name. `new_people` has no `bmi` or `overweight` column, so those cells are `NaN` for the new rows. If you set `join="inner"`, only columns present in *every* table are kept:

```python
pd.concat([people, new_people], join="inner")
```

`concat` keeps the original index labels. If they are not meaningful you can ask pandas to renumber the rows with `ignore_index=True`:

```python
pd.concat([people, new_people], ignore_index=True)
```

To concatenate *horizontally* (adding columns) set `axis=1`. The tables are aligned on their index, so darwin gets a row with `NaN` for everything that `people` does not know about him:

```python
pd.concat([people, residents.set_index("name")], axis=1)
```

This looks a lot like a `FULL OUTER JOIN` on the index, which is exactly what it is.

<div style="background:#fff8e1;border-left:6px solid #ffb300;padding:8px 12px;border-radius:4px;color:#000;">
<b>💡 Tip:</b> If you concatenate tables that share index labels you end up with duplicate labels, and <code>everyone.loc["bob"]</code> would return several rows. Pandas tolerates this, but it is usually a sign that you wanted <code>ignore_index=True</code> or a <code>merge</code>.
